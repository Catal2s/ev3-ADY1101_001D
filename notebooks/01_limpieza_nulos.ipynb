## Parte 1: Exploración y Tratamiento de Datos Faltantes
En esta fase inicial, prepararemos los datos del e-commerce público de Olist. El objetivo es identificar la cantidad y distribución de valores faltantes en tablas clave (Órdenes y Productos) para aplicar técnicas de imputación o eliminación, garantizando la consistencia de los datos para los modelos de Machine Learning.

```python
# Importación de librerías base
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

# Configuración visual
plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 12
sns.set_style("whitegrid")
sns.set_palette("husl")

print("✅ Librerías cargadas")
```

```python
# Carga de datos desde la carpeta raw
df_orders = pd.read_csv('../data/01_raw/olist_orders_dataset.csv')
df_products = pd.read_csv('../data/01_raw/olist_products_dataset.csv')

# Conversión de columnas de fecha al formato correcto
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'])

print(f"Total Órdenes: {df_orders.shape[0]:,}")
print(f"Total Productos: {df_products.shape[0]:,}\n")

# Mostrar columnas con nulos
print("--- NULOS EN ÓRDENES ---")
nulos_o = df_orders.isnull().sum()
print(nulos_o[nulos_o > 0])

print("\n--- NULOS EN PRODUCTOS ---")
nulos_p = df_products.isnull().sum()
print(nulos_p[nulos_p > 0])
```
